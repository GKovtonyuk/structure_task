```python
n = int(input("Кількість книг: "))
books = []

for i in range(n):
    book = {
        "author": input("Автор: "),
        "title": input("Назва: "),
        "year": int(input("Рік: ")),
        "kilkist": int(input("Кількість: ")),
        "price": int(input("Ціна: "))
    }
    books.append(book)

# Виводимо книги дешевші за 200 грн
s = 0
print("\nКниги дешевші за 200 грн:")
for book in books:
    if book["price"] < 200:
        print(f"{book['author']}, {book['title']}, {book['price']} грн")
        s += book["kilkist"] * book["price"]

print("Загальна сума вартості дешевих книг:", s)
```

```python
class Book: # створення структури
    pass  # тимчасово порожній клас

n = int(input("Кількість книг: "))
books = []

for i in range(n):
    b = Book() # створення об’єкта
    b.author = input("Автор: ") # додаємо значення полів b.author з клавіатури
    b.title = input("Назва: ")
    b.year = int(input("Рік: "))
    b.kilkist = int(input("Кількість: "))
    b.price = int(input("Ціна: "))
    books.append(b)

# Вивести книги, дешевші за 200 грн, і знайти суму
s = 0
print("\nКниги дешевші за 200 грн:")
for b in books:
    if b.price < 200:
        print(f"{b.author}, {b.title}, {b.price} грн")
        s += b.kilkist * b.price
print("Загальна вартість:", s)
```
